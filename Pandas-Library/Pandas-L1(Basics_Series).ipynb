## Creating a series in pandas: 
### Series in pandas are like arrays in numpy except the fact that we can change the index from default (0,1,2...) to anything (strings, float, int) as we want and can also store not only numbers but basically anything as the data corresponding to those index
---

```python
# Crating references :
my_list = ["a","b","c","d"]
my_array = np.array([1,2,3,4])
my_tuple = ("Dog","Cat","Bird","Frog")
my_dict = {"x" : "Best","Y" : "***" , "Z" : "Not_Human"}
print(my_list,my_array,my_tuple,my_dict, sep = "\n")
```

```python
# Creating Series:
# pd.Series(data , index , name , dtype, copy) -> Takes all these parameters , but mainly we will for now focus no data and index
# pd.Series(data = actual data that will be store , index = we can costomize index in which our data will be stored (default= 0,1,2,....))
# Series stores data in a vertical way!!!

s1 = pd.Series(data=my_array, index=my_list)     # here our data in my_array will be stored in according to the index in my_list [ Eg: at idx[a] -> 1 and so on]
print(s1)

print(s1["a"])            # In the series s1 at idex a(here) , 1 is stored

# print(s1[0])            # KeyError -> we can't do normal indexing now as we have changed the index but we can do normal slicing

s1["b"] = 69
print(s1)                # Series are mutable like lists (we can change the element at a certain index)

print(s1["a":"c"])       # We can also do slicing here but both start idx and end idx are included in case we use our custo non int
print(s1[1:2])           # we also can do normal slicing
```

```python
s2 = pd.Series(data = my_tuple,index = my_array)
print(s2)
print(s2[1])             # Indexing changes but slicing doesn't
print(s2[0:3])           # here the problem is we have made our custom index from 1, but normal slicing can be done from 0, hence this overlapping 

# Note : while giving custom index the len(data) must be equal to len(index) . If either of them is longer-> error
```

```python
# A series without pre-defined indexs' , the data will be stored in default indexs' (0,1,2.....)

s3 = pd.Series(data = my_tuple)
print(s3)

print(s3[0])            
print(s3[0:2])     # So in this case our normal idexing and slicing works
```

```python
# We can just pass data and index side by side without metioning data= and index=

s4 = pd.Series(my_tuple,my_list)          # first goes data and then goes index
print(s4)
print(s4["a"])
```

```python
print(my_dict)
```

```python
# In dictionaries as we use key value pairs we don't need to specify an index, the keys will be used as index and values as data(normally).
# So we just need to pass the dictionary in place of data

s5 = pd.Series(my_dict)        # keys-->index , values-->data
print(s5)

s6 = pd.Series(data = my_dict)     # Same as s5
print(s6)

s7 = pd.Series(index = my_dict)   # only keys will get printed as we have passed in place of index, the data will be as NaN
print(s7)
```

## Data that a pandas Series can hold:
### A pandas series can hold a variety of data types as data, unlike numpy arrays which could only hold int and float
---


```python
# Storing strings as data:

labels = ["a","b","c","d","e"]
print(pd.Series(data = labels))
```

```python
# Storing build-in functions as data
print(pd.Series(data = [len,sum,print,input,range] , index = labels))
```

```python
# Storing boolean:
print(pd.Series(data= [True,False]))
```

---
### Using an index
---

```python
# Basically indexing that we have already done:

ser1 = pd.Series(data = [1,2,3,4] , index = ["India","U.S.A","Germany","USSR"])
print(ser1)           
ser2 = pd.Series(data = ["India","U.S.A","Italy","USSR"] , index=[1,2,3,4])
print(ser2)
ser3 = pd.Series(data = [1,2,3,4] , index = ["India","U.S.A","Japan","USSR"])
print(ser3)
```

```python
print(ser1["India"])      # for labeled index we have to pass a string to get the data at that index
print(ser2[2])            # but for intiger index we can just pass the intiger
```
