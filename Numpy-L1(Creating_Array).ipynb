### (i) arange method
---

```python
# Creating 1D array with arange:
# np.arrange (start(default=0), upto stop, step(default=1))

arr = np.arange(12)               # creates a 1D array from 0 to 11
print(arr)

arr1 = np.arange(3,18)            # creates a 1D array from 0 to 17
print(arr1)

arr2 = np.arange(0,11,2)          # create a 1D array from 0 to 10 skipping every alt number
print(arr2)                       # output -> [ 0  2  4  6  8 10]
```

```python
# Creating 2D array with arange and reshape

arr = np.arange(12).reshape(3,4)      # reshape must include parameteres that product to the no of elements in array (i.e 3x4=12 -> here)
print(arr)

arr1 = np.arange(12).reshape(6,2)    # works perfectly as 6x2=12
print(arr1)

# arr2=np.arange(12).reshape(4,4)        # gives error as 4x4 != 12
# print(arr2)

# If you do not want to calculate the math for one of the dimensions, we can pass -1 into reshape. NumPy will automatically calculate the correct size for that dimension

arr3 = np.arange(0,200).reshape(5,-1)     # we have to pass atleast one parameter in reshape which will be sivisible by no of elements in arange  
arr3
```

### (iii) Linspace method
---

```python
# Return evenly spaced numbers in a 1D array over a specific interval

# np.linspace(start,stop,num)

#np.linspace(start, stop(included), no of evenly spaced elements in that start-stop interval(default=50))--> we must pass start, stop as argument

arr1 = np.linspace(0,100,11)   #-> it will start from 0 and go to 100 by printing 11 evenly spaced numbers inclusing 0 and 100
print(arr1)                    #-> returns an 1D array

arr2 = np.linspace(0,98)      # here only start and stop arg. are passed, default specific value is 50
print(arr2)

arr3 = np.linspace(0,200,60)    #-> fractional no will get printed
print(arr3)                     # -> 1D array
```

(a): rand: all no. between 0 to 1 have the equal probability of appearing


```python
# np.random.rand(order) : it populates an array (may be 1D or 2D) with random smaples in interval [0,1)-> 1 not included
# Order of this 1D or 2D array is decided by just writing the order without extra parenthesis --> 1D == np.random.rand(x) ; 2D == np.random.rand(y,x) or (x,1)
# This method will give a new array every time we run it with random samples in b/w 0 and 1

# Creating a 1D array of random samples:

vector1 = np.random.rand(5)                # creates a 1D array with 5 elements that is 5 random samples within interval [0,1)
print(vector1)                             # output -> [0.82283039 0.54720919 0.18223984 0.86635377 0.53787748]

vector2 = np.random.rand(10)               # creates a 1D array with 10 elements that is 10 random samples within interval [0,1)
print(vector2)

# without argument -> default 1
vector3 = np.random.rand()                 # default of order is 1, so it will create a random sample each time we run it if we haven't passed an argument
print(vector3)

vector4 = np.random.rand(0)               # creates an empty 1D array
print(vector4)                            # output : []
 
```

### (ii) Max, min, argmax, argmin functions
---

```python
# Max function and argmax : .max() returns the max element by value of a 1D or 2D array, if there are multiple max values it only returns the first occurance of the value
# argmax returns the index of the max value in the array : for multiple max values, returns the  index of occurance of the max value(index starts from 0)

# For 1D arrays:
v1 = np.arange(2,22)               # The max value will be 21
print(v1)
print(v1.max())                    # max() doesn't take any argument
print(v1.argmax())                 # returns the index of 21 in the array-> same as list index

v2 = np.random.rand(8)                # gives the max value of array 
print(v2)
print(v2.max())
print(v2.argmax())                    # returns the index of the max value

v3 = np.random.randint(10,19,20)            # in this specific run 18 is the highest no and 18 has occured thrice
print(v3)
print(v3.max())
print(v3.argmax())                          # returns the index of first occurance of max value 
```

```python
# For 2D array: 
# argmax index for 2D array is counted from the top left element row-wise(horizontally) till the last bottom right element

matrix1 = np.random.rand(5,5)
print(matrix1)
print(matrix1.max())
print(matrix1.argmax())
```

```python
# Min function: Gives the minimium value
# argmin-> the the index of first occurance of minimum value; logic same as argmax

# For 1D array
vec1 = np.random.randint(0,12,25)              # multiple min values
print(vec1)
print(vec1.min)                               # gives object location
print(vec1.min())
print(vec1.argmin())                          # returns index of occurance of first min value in case of multiplr


# For 2D array
ma1 = np.random.randn(5,5)
print(ma1)
print(ma1.min())                   # minimum value is negative in this case
print(ma1.argmin())                # for 2D array index same logic as argmax-> topleft to bottom right row-wise
```

### 🧠 Understanding NumPy Integer Datatypes and Memory Capacity

In this context, the number of **bits** determines the minimum and maximum whole numbers that can be stored in each slot of a NumPy array. Choosing the right size helps optimize computer memory usage.

---

#### 📊 Bit Capacity Breakdown

| Datatype | Memory per Element | Minimum Value | Maximum Value | Best Used For |
| :--- | :--- | :--- | :--- | :--- |
| **`int8`** | 1 Byte (8 bits) | $-128$ | $127$ | Very small numbers (e.g., ages, days of the month) |
| **`int16`** | 2 Bytes (16 bits) | $-32,768$ | $32,767$ | Medium numbers (e.g., daily temperature records) |
| **`int32`** | 4 Bytes (32 bits) | $-2,147,483,648$ | $2,147,483,647$ | Standard calculations (e.g., city populations) |
| **`int64`** | 8 Bytes (64 bits) | $-9,223,372,036,854,775,808$ | $9,223,372,036,854,775,807$ | Massive values (e.g., financial ledger totals, scientific data) |

---

#### 🔍 What is Included in this Context?

1. **The Math Formula:** The total numbers a bit size can hold is calculated by $2^{\text{bits}}$. For signed integers (which support both negative and positive numbers), the formula for the maximum value is $2^{(\text{bits} - 1)} - 1$.
2. **Memory Efficiency:** Smaller bit sizes (like `int8`) use **8 times less memory** than larger sizes (like `int64`).
3. **The Overflow Risk:** If a calculation produces a number larger than the maximum capacity of its bit size, the data will **overflow**, resulting in broken or corrupted negative numbers.
---
### 🖥️ Default Bit Sizes: 32-bit vs. 64-bit Arrays

When you don't explicitly specify a `dtype`, NumPy automatically assigns a default bit size depending on your computer's operating system (OS) and the size of your numbers.

---

#### 📌 Which Array Becomes 32-bit?

An array will default to **`int32`** in the following scenarios:

1. **Windows OS Defaults:** On standard **Windows** machines, NumPy defaults to creating `int32` arrays for normal integer values.
2. **Explicit Declaration:** When you force the array to use 32-bit using `dtype=np.int32`.
3. **Decimals/Floating Points (`float32`):** If you create an array of decimal numbers and choose lower-precision floating points to save space.

---

#### 📌 Which Array Becomes 64-bit?

An array will default to **`int64`** in the following scenarios:

1. **Linux and macOS Defaults:** On modern 64-bit **Linux** and **Apple Mac** computers, NumPy automatically defaults to `int64` for all general integers.
2. **Massive Numbers:** If an integer on *any* operating system is too large to fit in a 32-bit limit (outside the range of $-2$ billion to $+2$ billion), NumPy automatically bumps it up to `int64`.
3. **Default Floating Points (`float64`):** When you create an array with decimals (e.g., `[1.5, 2.3, 4.0]`), NumPy defaults to 64-bit precision across all operating systems.


```python
import numpy as np

# Create three arrays with the exact same 5 numbers, but different bit sizes
data = [10, 20, 30, 40, 50]

arr_int8 = np.array(data, dtype=np.int8)
arr_int32 = np.array(data, dtype=np.int32)
arr_int64 = np.array(data, dtype=np.int64)

# Print the total memory used by each array
print(f"📊 int8 array memory:  {arr_int8.nbytes} bytes")
print(f"📊 int32 array memory: {arr_int32.nbytes} bytes")
print(f"📊 int64 array memory: {arr_int64.nbytes} bytes")
```

```python
# Example of storage space in bits

# 1. Small bucket (int8) - Uses very little memory
# Best for tiny numbers between -128 and 127
small_array = np.array([10, 15, 12], dtype=np.int8)
print("Small Array Type:", small_array.dtype)

# 2. Medium bucket (int32) - Your default size
# Best for normal numbers up to 2 billion
medium_array = np.array([10, 15, 12], dtype=np.int32)
print("Medium Array Type:", medium_array.dtype)

# 3. Large bucket (int64) - Uses the most memory
# Best for giant numbers up to 9 quintillion
large_array = np.array([10, 15, 12], dtype=np.int64)
print("Large Array Type:", large_array.dtype)
```
